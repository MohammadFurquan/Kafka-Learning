
# <span style="color:blue">Schema Registry in Apache Kafka</span>

# What is a Schema?

A Schema defines the structure of data.

It describes:

✅ Field Names

✅ Data Types

✅ Required Fields

✅ Optional Fields

✅ Message Format

---

## Example

Data:

```json
{
  "id": 101,
  "name": "Arif"
}
```

Schema:

```json
{
  "id": "int",
  "name": "string"
}
```

The schema tells applications:

```text
What fields exist?

What data types are expected?
```

---

# Why Do We Need Schemas?

Without schemas:

```text
Producer and Consumer
must manually agree
on message structure.
```

This can lead to:

❌ Data Inconsistency

❌ Consumer Failures

❌ Difficult Maintenance

---

# What is Schema Registry?

Schema Registry is a centralized repository used to store, manage, validate, and version schemas used by Kafka messages.

### Simple Definition

```text
Schema Registry
=
Central Repository For Schemas
```

It stores schemas for:

```text
Avro

Protobuf

JSON Schema
```

---

# Why Do We Need Schema Registry?

Suppose Producer sends:

```json
{
  "id": 101,
  "name": "Arif"
}
```

Consumer understands this schema.

---

Later Producer changes data:

```json
{
  "id": 101,
  "name": "Arif",
  "email": "arif@gmail.com"
}
```

Now:

```text
Producer Schema
      ≠
Consumer Schema
```

Possible Result:

```text
Consumer Failure
```

Schema Registry helps manage these changes safely.

---

# Problems Without Schema Registry

```text
Producer V1

Consumer V2
```

Different schemas.

Result:

```text
Compatibility Issues
```

---

Challenges:

❌ No Central Control

❌ Schema Drift

❌ Difficult Version Management

❌ Breaking Changes

❌ Consumer Failures

---

# How Schema Registry Works

## Producer Side

Producer wants to send:

```json
{
  "id": 101,
  "name": "Arif"
}
```

Before sending:

```text
Register Schema
```

with Schema Registry.

---

## Schema Registration

Schema:

```json
{
  "id": "int",
  "name": "string"
}
```

Schema Registry assigns:

```text
Schema ID = 25
```

---

## Kafka Message

Producer does not send the full schema every time.

Instead:

```text
Schema ID + Data
```

Example:

```text
Schema ID = 25

Data = Customer Record
```

---

## Consumer Side

Consumer receives:

```text
Schema ID = 25
```

Consumer requests:

```text
Schema 25
```

from Schema Registry.

---

Consumer Flow:

```text
Schema ID
      ↓
Fetch Schema
      ↓
Deserialize Data
      ↓
Original Object
```

# What is Schema Evolution?

Schema Evolution is the ability to modify schemas over time while maintaining compatibility between producers and consumers.

### Simple Definition

```text
Schema Evolution
=
Changing Schemas Safely Over Time
```

---

# Example

## Version 1

```json
{
  "id": "int",
  "name": "string"
}
```

---

## Version 2

```json
{
  "id": "int",
  "name": "string",
  "email": "string"
}
```

Schema evolved by adding a new field.

---

# Why Is Schema Evolution Needed?

Applications evolve.

Example:

```text
Customer System
```

Initially:

```text
ID
Name
```

Later:

```text
ID
Name
Email
Phone
```

Schema Registry allows these changes safely.

---

# Schema Evolution Rules ⭐⭐⭐

## Safe Changes

### Add Optional Fields

Example:

```json
{
  "id": "int",
  "name": "string",
  "email": "string"
}
```

---

### Add Fields With Default Values

Example:

```json
{
  "country": "India"
}
```

Consumers can still process data.

---

## Risky Changes

### Remove Required Fields

Example:

```json
Remove:
name
```

Existing consumers may fail.

---

### Change Data Types

Example:

```text
id

int → string
```

Can break compatibility.

---

# Schema Compatibility Modes ⭐⭐⭐⭐

Schema Registry provides multiple compatibility modes.

---

## 1. BACKWARD

```text
New schema can read data written by the old schema.
```

Example:

Consumer upgraded.

Producer still uses old schema.

Works successfully.

---

## 2. FORWARD

```text
Old schema can read data written by the new schema.
```

Example:

Producer upgraded.

Consumer still uses old schema.

Works successfully.

---

## 3. FULL

```text
Both backward and forward compatible.
```

Supports:

✅ Backward Compatibility

✅ Forward Compatibility

---

## 4. NONE

```text
No validation is performed.
```

Any schema change is accepted.

Risk:

```text
Breaking Changes Possible
```

---

# Compatibility Diagram

```text
BACKWARD

Old Data
    ↓
New Schema
```

---

```text
FORWARD

New Data
    ↓
Old Schema
```

---

```text
FULL

Old ↔ New
```

---

```text
NONE

No Compatibility Checks
```
