# <span style="color:blue">KRaft vs ZooKeeper</span>

# What is ZooKeeper?

ZooKeeper is a distributed coordination service that was used by Apache Kafka to manage cluster metadata, leader election, broker registration, and cluster coordination.

In older Kafka architectures, ZooKeeper acted as the cluster manager.

### Simple Definition

```text
ZooKeeper
=
External Cluster Manager For Kafka
```

---

# Why Was ZooKeeper Needed?

Kafka brokers need a central system to manage:

✅ Broker Information

✅ Topic Metadata

✅ Leader Election

✅ Partition Information

✅ Cluster State

ZooKeeper performed all these tasks.

---

# What is KRaft?

KRaft stands for:

```text
Kafka Raft Metadata Mode
```

KRaft is Kafka's built-in metadata management and cluster coordination system.

It removes the need for ZooKeeper.

### Simple Definition

```text
KRaft
=
Built-In Kafka Cluster Manager
```

---

# Why Was KRaft Introduced?

Managing both:

```text
Kafka Cluster

+

ZooKeeper Cluster
```

made deployments more complex.

To simplify architecture, Kafka introduced KRaft.

Now Kafka handles:

✅ Metadata Management

✅ Leader Election

✅ Broker Registration

✅ Cluster Coordination

without ZooKeeper.

---


# <span style="color:blue">KRaft vs ZooKeeper</span>

## Why Do We Need a Cluster Manager in Kafka?

A Kafka cluster needs a component that can manage:

✅ Broker Registration

✅ Metadata Management

✅ Leader Election

✅ Cluster Coordination

✅ Failure Detection

✅ ISR Information

Without a cluster manager, brokers would not know:

```text
Which broker is alive?

Who is the leader?

Which partitions exist?

What is the cluster state?
```

---

# <span style="color:green">ZooKeeper (Old Architecture)</span>

In older Kafka versions, Kafka depended on ZooKeeper for cluster management.

Architecture:

```text
                ZooKeeper
                     ▲
                     │
      ┌──────────────┼──────────────┐
      │              │              │
      ▼              ▼              ▼

   Broker1       Broker2        Broker3
```

Kafka relied on ZooKeeper for critical cluster information.

---

# Responsibilities of ZooKeeper

## 1. Broker Registration

When a broker starts:

```text
Broker1
```

it registers itself with ZooKeeper.

ZooKeeper knows:

```text
Which brokers are alive?
```

---

## 2. Metadata Management

ZooKeeper stores:

```text
Topics

Partitions

Leaders

ISR Information

Broker Information
```

---

## 3. Leader Election

If a Leader fails:

```text
Broker1 ❌
```

ZooKeeper helps determine:

```text
Who becomes the new Leader?
```

---

## 4. Cluster Coordination

ZooKeeper coordinates communication and state across brokers.

---

# Problems with ZooKeeper

Kafka architecture became more complex.

To run Kafka, organizations had to maintain:

```text
Kafka Cluster

+

ZooKeeper Cluster
```

Two separate distributed systems.

---

## Challenges

✅ Extra Infrastructure

✅ More Maintenance

✅ Higher Operational Cost

✅ More Failure Points

✅ More Monitoring

✅ More Complex Upgrades

# Controller Quorum ⭐

Modern Kafka uses:

```text
Controller Quorum
```

instead of ZooKeeper.

---

## What is Controller Quorum?

A group of controller nodes responsible for maintaining metadata consistency.

Example:

```text
Controller1

Controller2

Controller3
```

Together:

```text
Controller Quorum
```

---

# Why Controller Quorum?

Benefits:

✅ High Availability

✅ Fault Tolerance

✅ Metadata Consistency

✅ Reliable Leader Election

✅ Prevent Split Brain Problems
