
# <span style="color:blue">Kafka Connect</span>

# What is Kafka Connect?

Kafka Connect is a framework provided by Apache Kafka for reliably and scalably moving data between Kafka and external systems.

### Simple Definition

```text
Kafka Connect
=
Framework For Moving Data
Between Kafka And External Systems
```

---

# Why Do We Need Kafka Connect?

Without Kafka Connect:

```text
Write Producer Code

Write Consumer Code

Handle Retries

Handle Scaling

Handle Monitoring

Handle Failures

Handle Recovery
```

A lot of custom development and maintenance is required.

---

## With Kafka Connect

```text
Configuration Instead Of Code
```

Simply configure connectors and Kafka Connect handles the rest.

---

# Problems Solved By Kafka Connect

✅ Data Integration

✅ Fault Tolerance

✅ Scalability

✅ Monitoring

✅ Retry Handling

✅ Offset Management

✅ Distributed Processing

✅ Easy Connectivity

---

# What Systems Can Kafka Connect Integrate With?

Examples:

```text
MySQL

PostgreSQL

Oracle

MongoDB

Snowflake

Amazon S3

Elasticsearch

Redis

Data Lakes
```

---

# Kafka Connect Architecture

```text
External System
        ↓

Source Connector
        ↓

      Kafka
        ↓

Sink Connector
        ↓

External System
```

# What is a Connector?

A Connector defines the data integration.

### Simple Definition

```text
Connector
=
Definition Of Data Movement
```

It specifies:

```text
Source

Destination

Configuration
```

---

# Types of Connectors

Kafka Connect provides two types of connectors:

```text
Source Connector

Sink Connector
```

---

# 1. Source Connector

A Source Connector reads data from an external system and writes it to Kafka.

---

## Flow

```text
External System
       ↓
Source Connector
       ↓
Kafka
```

---

## Example

```text
MySQL
   ↓
JDBC Source Connector
   ↓
Kafka
```

---

---

# 2. Sink Connector

A Sink Connector reads data from Kafka and writes it to an external system.

---

## Flow

```text
Kafka
   ↓
Sink Connector
   ↓
External System
```

---

## Example

```text
Kafka
   ↓
Snowflake Sink Connector
   ↓
Snowflake
```

# What is a Task?

A Task performs the actual data transfer work.

### Simple Definition

```text
Task
=
Unit Of Work
```

---

# Why Tasks?

Large workloads can be processed in parallel.

---

## Example

Connector:

```text
MySQL → Kafka
```

Kafka Connect creates:

```text
Task1

Task2

Task3
```

Each task processes part of the workload.

---

# Connector vs Task ⭐⭐⭐

## Connector

Defines:

```text
What Data To Move
```

Example:

```text
MySQL → Kafka
```

---

## Task

Defines:

```text
How Data Movement Is Executed
```


# <span style="color:blue">Kafka Connect: Standalone Mode vs Distributed Mode</span>

# What is a Worker?

A Worker is a Kafka Connect process responsible for running connectors and tasks.

### Simple Definition

```text
Worker
=
Kafka Connect Process
```

A Worker performs:

✅ Connector Execution

✅ Task Execution

✅ Data Movement

✅ Offset Management

✅ Failure Handling

---

# Kafka Connect Worker

Example:

```text
Worker1
```

running on:

```text
Machine1
```

This worker executes connectors and tasks.

---

# Standalone Mode

## What is Standalone Mode?

Standalone Mode runs Kafka Connect using:

```text
Only One Worker
```

---

## Architecture

```text
            Worker1

               ↓

        Source Connector

               ↓

            Task1
```

---

## Example

Pipeline:

```text
MySQL
   ↓
Kafka
```

Only one worker exists:

```text
Worker1
```

Worker1 performs:

```text
Read MySQL

Write To Kafka

Manage Tasks
```

Everything runs inside a single worker.

---

## Failure Scenario

Suppose:

```text
Worker1 Crashes ❌
```

Result:

```text
Entire Pipeline Stops ❌
```

Because there is no backup worker available.

---

## Advantages

✅ Simple Setup

✅ Easy To Configure

✅ Low Resource Usage

✅ Good For Learning

---

## Limitations

❌ No High Availability

❌ No Fault Tolerance

❌ Limited Scalability

❌ Not Recommended For Production

---

## Use Cases

```text
Development

Testing

Learning

Small Workloads
```

---

# Real Life Analogy

Imagine a shop with:

```text
1 Employee
```

That employee handles:

```text
Billing

Packing

Delivery
```

If the employee leaves:

```text
Shop Stops Working
```

This is similar to:

```text
Standalone Mode
```

---

# Distributed Mode ⭐⭐⭐

## What is Distributed Mode?

Distributed Mode runs Kafka Connect using:

```text
Multiple Workers
```

that work together as a cluster.

---

## Example

```text
Worker1

Worker2

Worker3
```

---

## Architecture

```text
        Kafka Connect Cluster

     ┌────────┬────────┬────────┐

     ▼        ▼        ▼

  Worker1  Worker2  Worker3
```

Together these workers form:

```text
Kafka Connect Cluster
```

---

# Task Distribution

Suppose a connector creates:

```text
Task1

Task2

Task3
```

Kafka Connect distributes tasks.

Example:

```text
Worker1 → Task1

Worker2 → Task2

Worker3 → Task3
```

Workload is shared across multiple workers.

---

# Failure Scenario

Suppose:

```text
Worker2 Crashes ❌
```

Task2 was running on Worker2.

Kafka Connect automatically reassigns Task2 to:

```text
Worker1

OR

Worker3
```

Result:

```text
Pipeline Continues ✅
```

---

# Why Distributed Mode?

Because:

```text
More Workers
      ↓
More Parallelism
      ↓
Better Performance
```

---

## Example

Workload:

```text
1 Million Records
```

---

### Standalone Mode

```text
1 Worker
```

Processes everything alone.

Result:

```text
Slower
```

---

### Distributed Mode

```text
Worker1

Worker2

Worker3
```

All workers process data in parallel.

Result:

```text
Faster
```

---

# Benefits of Distributed Mode

✅ High Availability

✅ Fault Tolerance

✅ Load Balancing

✅ Scalability

✅ Better Performance

✅ Production Ready

---

# Use Cases

```text
Production Systems

Large Data Pipelines

CDC Pipelines

Enterprise Workloads
```
