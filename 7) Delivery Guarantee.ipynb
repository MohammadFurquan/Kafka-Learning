
# <span style="color:blue">Kafka Delivery Guarantees</span>

## What are Delivery Guarantees?

Delivery Guarantees define how reliably Kafka delivers messages from Producer to Consumer.

They answer the question:

```text
How many times can a message be delivered?
```

Kafka supports three delivery guarantees:

```text
1. At Most Once

2. At Least Once

3. Exactly Once
```

---

# End-To-End Flow

```text
Producer
    ↓
Kafka
    ↓
Consumer
```

Depending on failures and offset commits, Kafka can provide:

```text
At Most Once

At Least Once

Exactly Once
```

---

# Why Do Delivery Guarantees Matter?

Suppose a Producer sends:

```text
Order Created
```

Consumer receives it.

Now imagine:

```text
Consumer Crash

Network Failure

Broker Failure

Offset Commit Failure
```

Questions:

```text
Can the data be lost?

Can the message be processed twice?

Can Kafka guarantee exactly one delivery?
```

These questions are answered by Delivery Guarantees.

---

# 1. At Most Once

## Definition

A message is delivered:

```text
0 or 1 Time
```

This means:

```text
Data Loss Possible

Duplicates Not Possible
```

---

## How It Works?

Consumer commits offset before processing.

Flow:

```text
Read Message
      ↓
Commit Offset
      ↓
Process Message
```

---

## Failure Scenario

```text
Read Message
      ↓
Commit Offset
      ↓
Consumer Crash ❌
```

After restart:

```text
Kafka thinks message is already processed
```

Message is skipped.

Result:

```text
Data Lost
```

---

## Diagram

```text
Message
   ↓

Consumer

   ↓

Offset Commit ✅

   ↓

Crash ❌
```

Result:

```text
Message Lost
```

---

## Characteristics

✅ No Duplicate Processing

❌ Data Loss Possible

✅ Fast Processing

---

## Use Cases

```text
Application Logs

Metrics

Monitoring Events
```

Where losing a few messages is acceptable.

---

# 2. At Least Once ⭐

## Definition

A message is delivered:

```text
1 or More Times
```

This means:

```text
No Data Loss

Duplicates Possible
```

---

## How It Works?

Consumer commits offset after successful processing.

Flow:

```text
Read Message
      ↓
Process Message
      ↓
Commit Offset
```

---

## Failure Scenario

```text
Read Message
      ↓
Process Message ✅
      ↓
Crash ❌
      ↓
Offset Not Committed
```

Consumer restarts.

Kafka thinks:

```text
Message Not Processed
```

Kafka sends message again.

Result:

```text
Duplicate Processing
```

---

## Diagram

```text
Message
   ↓

Consumer

   ↓

Process ✅

   ↓

Crash ❌

(No Offset Commit)
```

After Restart:

```text
Read Same Message Again
```

Result:

```text
Duplicate
```

---

## Characteristics

✅ No Data Loss

✅ Reliable

❌ Duplicates Possible

---

## Why Is It Popular?

Most businesses prefer:

```text
Duplicate Data
```

over

```text
Lost Data
```

Duplicates can be cleaned later.

Lost business data cannot be recovered.

---

## Real Data Engineering Example

```text
Kafka
   ↓
Spark
   ↓
Snowflake
```

Most pipelines operate using:

```text
At Least Once
```

# 3. Exactly Once ⭐⭐⭐

## Definition

A message is delivered:

```text
Exactly One Time
```

This means:

```text
No Data Loss

No Duplicates
```

---

## Goal

```text
Process Every Message

Exactly Once
```

No more.

No less.

---

## How Kafka Achieves Exactly Once?

Using:

```text
Idempotent Producer
+
Transactions
```

---

## Idempotent Producer

Prevents duplicate writes from Producer retries.

Example:

```text
Send Order101
```

Retry occurs.

Kafka recognizes duplicate send.

Only one record is stored.

---

## Transactions

Ensure multiple writes succeed or fail together.

```text
All Success ✅

OR

All Fail ❌
```

No partial processing.

---

## Characteristics

✅ No Data Loss

✅ No Duplicates

✅ Highest Accuracy

❌ Higher Complexity

❌ More Overhead

---

## Use Cases

```text
Banking Systems

Financial Transactions

Payment Platforms

Inventory Systems
```


# <span style="color:blue">Consumer Offset Commit Timing and Delivery Guarantees</span>

## Why Do Delivery Guarantees Exist?

The main reason Delivery Guarantees exist is because of:

```text
Consumer Offset Commit Timing
```

The timing of when a Consumer commits its offset determines whether:

```text
Data Can Be Lost

OR

Duplicate Processing Can Happen
```

---

# What Is Offset Commit?

A Consumer reads messages from Kafka.

Example:

```text
Offset 0 → Order1

Offset 1 → Order2

Offset 2 → Order3
```

After processing messages, the Consumer informs Kafka:

```text
I have successfully processed data up to this offset.
```

This process is called:

```text
Offset Commit
```

Example:

```text
Processed:

Offset 0
Offset 1
```

Kafka Stores:

```text
Committed Offset = 1
```

---

# Scenario 1: Commit BEFORE Processing

## Flow

```text
Read Message
      ↓
Commit Offset
      ↓
Process Message
```

---

## Example

Message:

```text
Offset 5 → Order5
```

Consumer:

```text
Read Order5 ✅
```

Immediately commits:

```text
Offset 5 Committed ✅
```

Now Kafka believes:

```text
Order5 Successfully Processed
```

But processing has not actually happened yet.

---

## Failure Scenario

```text
Read Order5 ✅
      ↓
Commit Offset ✅
      ↓
Application Crash ❌
```

---

## What Happens After Restart?

Kafka checks committed offset.

```text
Committed Offset = 5
```

Kafka assumes:

```text
Order5 Already Processed
```

Consumer starts from:

```text
Offset 6
```

---

## Result

```text
Order5 Lost ❌
```

---

## Diagram

```text
Order5
   ↓

Read ✅

   ↓

Commit Offset ✅

   ↓

Crash ❌

   ↓

Restart

   ↓

Order6
```

Order5 is never processed.

---

## Delivery Guarantee

```text
At Most Once
```

---

## Characteristics

✅ No Duplicates

❌ Data Loss Possible

---

# Interview Questions & Answers

## Q1. What Is Offset Commit?

Offset Commit is the process of storing the last successfully processed offset in Kafka.

---

## Q2. Why Is Offset Commit Important?

It helps Kafka track consumer progress and resume processing after failures.

---

## Q3. What Happens If Offset Is Committed Before Processing?

Data loss can occur if the application crashes before processing is completed.

---

## Q4. Which Delivery Guarantee Commits Offset Before Processing?

```text
At Most Once
```

---

## Q5. What Happens If Offset Is Committed After Processing?

Duplicate processing can occur if the application crashes before the offset is committed.

---

## Q6. Which Delivery Guarantee Commits Offset After Processing?

```text
At Least Once
```

---

## Q7. Why Does At Most Once Cause Data Loss?

Because Kafka believes the message is already processed even if processing never happened.

---

## Q8. Why Does At Least Once Cause Duplicates?

Because processing may succeed but offset commit may fail.

---

## Q9. Which Is Better For Data Engineering: Data Loss Or Duplicates?

```text
Duplicates
```

are usually preferred over data loss.
