
# <span style="color:blue">Replication</span>

## What is Replication?

Replication means creating multiple copies of a Kafka Partition across different brokers.

### Simple Definition

```text
Replication
=
Multiple copies of a partition
```

Purpose:

```text
Fault Tolerance
```

---

## Why Do We Need Replication?

Suppose:

```text
Broker1
```

contains all your data.

Suddenly:

```text
Broker1 Crashes
```

Without replication:

```text
Data Lost ❌
```

With replication:

```text
Broker2 ✅

Broker3 ✅
```

Data remains available.

---

---

```text

```

```text

```

Data Safe ✅

# <span style="color:blue">Replication Factor (RF)</span>

Replication Factor defines how many copies of a partition should exist.

---

## Example

```text
RF = 3
```

Means:

```text
1 Original Copy

2 Replica Copies
```

Total:

```text
3 Copies
```

---

## Example

Topic:

```text
Orders
```

Partition:

```text
Partition-0
```

Replication Factor:

```text
RF = 3
```

Storage:

```text
Broker1 → Copy 1

Broker2 → Copy 2

Broker3 → Copy 3
```

---

## Important Rule

Replication Factor can never be greater than broker count.

Example:

```text
3 Brokers
```

Valid:

```text
RF = 1
RF = 2
RF = 3
```

Invalid:

```text
RF = 4
```

# <span style="color:blue">Leader-Follower Architecture</span>

In Kafka, every partition has:

```text
1 Leader

0 or More Followers
```

---

## Example

```text
Partition-0

Broker1 → Leader

Broker2 → Follower

Broker3 → Follower
```

---

# <span style="color:green">Leader-Follower Diagram</span>

```text
                   Producer
                       |
                       v

              Broker1 (Leader)
                    /    \
                   /      \
                  v        v

      Broker2(Follower)  Broker3(Follower)
```

---

## What is Leader?

Leader is the main replica that handles:

✅ Writes

✅ Reads

✅ Client Requests

---

## What is Follower?

Follower maintains a synchronized copy of the Leader.

Follower continuously copies data from Leader.

Example:

```text
Leader

Order1
Order2
Order3
```

Followers:

```text
Order1
Order2
Order3
```

continuously synchronize.

---

# Producer Write Flow

Producer always sends data to:

```text
Leader
```

Example:

```text
Producer
   |
   v

Leader
```

Never:

```text
Producer → Follower ❌
```

---

# Consumer Read Flow

Normally:

```text
Consumer
    |
    v

Leader
```

Consumers usually read from Leader.

---

# Leader Failure Scenario

Before Failure:

```text
Broker1 → Leader

Broker2 → Follower

Broker3 → Follower
```

---

Broker1 Crash

```text
Broker1 ❌
```

Kafka elects new leader.

After Election:

```text
Broker2 → Leader

Broker3 → Follower
```

---

# Failover Diagram

Before:

```text
Partition-0

Broker1 → Leader

Broker2 → Follower

Broker3 → Follower
```

After Crash:

```text
Partition-0

Broker2 → Leader

Broker3 → Follower
```

Processing continues.

---

# Benefits of Leader-Follower Model

✅ No Downtime

✅ High Availability

✅ Automatic Failover

✅ Continuous Processing

✅ Disaster Recovery

---

# <span style="color:blue">Retention</span>

## What is Retention?

Retention means Kafka keeps messages for a configured period even after they are consumed.

This is one of Kafka's most powerful features.

---

# Traditional Queue vs Kafka

Traditional Queue:

```text
Message Read
      ↓
Message Deleted
```

Kafka:

```text
Message Read
      ↓
Message Still Stored
```

---

# Example

Retention:

```text
7 Days
```

Producer sends message:

```text
Order1
```

Consumer reads:

```text
Order1
```

Kafka still stores it:

```text
7 Days
```

---

# Retention Flow Diagram

```text
Day 1
Message Created

        ↓

Day 2
Consumer Reads

        ↓

Day 3
Message Still Available

        ↓

Day 7
Message Still Available

        ↓

Retention Expired
Message Deleted
```

---

# Why Do We Need Retention?

## Replay

Consumer can read old messages again.

---

## Backfill

New consumer can process historical data.

---

## Audit

Historical events can be reviewed.

---

## Recovery

Recover after application failures.

---

# Replay Example

Topic:

```text
Orders
```

Messages:

```text
Offset 0

Offset 1

Offset 2

Offset 3
```

Consumer reads all messages.

2 Days Later:

Consumer resets offset.

```text
Offset = 0
```

Reads all data again.

---

# Retention Types

## Time Based Retention

Most Common

Examples:

```text
7 Days

14 Days

30 Days
```

---

## Size Based Retention

Kafka retains data until storage limit is reached.

Examples:

```text
100 GB

500 GB

1 TB
```

Old messages are deleted when the limit is reached.

---

# <span style="color:green">1. Replay</span>

## What is Replay?

Replay means reading old Kafka messages again.

### Example

Consumer currently at:

```text
Offset = 1000
```

Offset reset:

```text
Offset = 0
```

Consumer starts reading again:

```text
Offset 0
Offset 1
Offset 2
...
```

This is called Replay.

---

## Purpose of Replay

✅ Read historical messages again

✅ Reprocess old events

✅ Testing

✅ Debugging

✅ Analytics

---

## Replay Diagram

```text
Kafka Topic

Offset 0
Offset 1
Offset 2
Offset 3
Offset 4
```

Consumer:

```text
Read Till Offset 4
```

Later:

```text
Reset Offset = 0
```

Read again:

```text
Offset 0
Offset 1
Offset 2
Offset 3
Offset 4
```

---

## Interview Definition

```text
Replay means reading historical Kafka messages again by resetting offsets.
```

---

# <span style="color:green">2. Backfill</span>

## What is Backfill?

Backfill means reprocessing historical data to populate missing data in downstream systems.

### Example

Architecture:

```text
Kafka
   ↓
Snowflake
```

Snowflake missed some records.

Now:

```text
Replay Old Data
       ↓
Load Missing Records
```

This process is called Backfill.

---

## Purpose of Backfill

✅ Fill Missing Data

✅ Populate New Systems

✅ Rebuild Analytics Tables

✅ Recover Missing Records

---

## Backfill Diagram

```text
Kafka
   ↓

Historical Data

   ↓

Replay

   ↓

Snowflake

(Missing Records Filled)
```

---

## Real World Example

A new analytics table is created.

Need:

```text
Last 6 Months Data
```

Historical Kafka data is processed again.

```text
Kafka
   ↓
Replay
   ↓
Analytics Table
```

This is Backfill.

---

## Interview Definition

```text
Backfill means reprocessing historical data to populate missing records or load data into a new downstream system.
```

---

# <span style="color:green">3. Recovery</span>

## What is Recovery?

Recovery means resuming processing after a failure.

---

## Example

```text
Offset 0
Offset 1
Offset 2
Offset 3
```

Kafka stored:

```text
Consumer Offset = 3
```

Suddenly:

```text

```

After restart:

```text
Continue From Offset 4
```

This is Recovery.

---

## Purpose of Recovery

✅ Continue Processing

✅ Prevent Data Loss

✅ Resume After Failure

✅ Reliable Pipelines

---

## Recovery Diagram

```text

Continue From Offset 4
```

---

## Interview Definition

```text
Recovery means restoring processing after a failure using saved consumer offsets.
```
