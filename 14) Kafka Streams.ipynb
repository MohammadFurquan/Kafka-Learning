
# <span style="color:blue">Kafka Streams</span>

# What is Kafka Streams?

Kafka Streams is a client library provided by Apache Kafka for building real-time stream processing applications.

It allows applications to:

✅ Read Data From Kafka

✅ Process Data

✅ Transform Data

✅ Aggregate Data

✅ Join Data

✅ Write Results Back To Kafka

---

## Simple Definition

```text
Kafka Streams
=
Read Data From Kafka
      ↓
Process Data
      ↓
Write Results Back To Kafka
```

---

# Why Do We Need Kafka Streams?

Suppose Kafka contains an orders topic.

```text
Order1 = 100

Order2 = 200

Order3 = 300
```

Business asks:

```text
Calculate Total Revenue
```

Kafka itself only stores events.

Kafka cannot perform calculations.

We need:

```text
Processing
```

This is where Kafka Streams comes in.

# What is a Stream?

A stream is a continuous flow of events.

Example:

```text
Order Created

Order Created

Order Updated

Order Cancelled

Order Delivered
```

Events continuously arrive.

---

# Kafka Streams Core Concepts

Two very important concepts:

```text
KStream

KTable
```

---

# What is KStream?

KStream represents a stream of events.

Every event is preserved.

---

## Example

```text
Order1 = Created

Order1 = Updated

Order1 = Updated
```

All events exist.

---

## Simple Definition

```text
KStream
=
Event History
```

---

# KStream Diagram

```text
Order1 → Created

Order1 → Updated

Order1 → Updated

Order2 → Created
```

All events remain.

---

# What is KTable?

KTable represents the latest state for each key.

---

## Example

```text
Customer1 → Arif

Customer1 → Mohammad Arif

Customer1 → M Arif
```

KTable stores:

```text
Customer1 → M Arif
```

Only latest value matters.

---

## Simple Definition

```text
KTable
=
Current State
```

---

# KTable Diagram

Before:

```text
Customer1 → Arif

Customer1 → Mohammad Arif

Customer1 → M Arif
```

After KTable View:

```text
Customer1 → M Arif
```

---

# KStream vs KTable ⭐⭐⭐

## KStream

```text
Stores All Events
```

Example:

```text
A

B

C
```

Every event retained.

---

## KTable

```text
Stores Latest Value Per Key
```

Example:

```text
Customer1 → Latest Value
```

---

# Comparison

## KStream

```text
Represents Event History
```

```text
Every Event Retained
```

```text
Append Only
```

---

## KTable

```text
Represents Current State
```

```text
Only Latest Value Matters
```

```text
State Based
```

---

# Relation With Log Compaction

You already learned:

```text
Log Compaction
=
Keep Latest Value Per Key
```

This is exactly how KTable treats data.

---

## Connection

```text
Compacted Topic
      ↓
KTable
```

Very common interview question.

# Stateful vs Stateless Processing

Another important Kafka Streams concept.

---

# What is Stateless Processing?

Each record is processed independently.

No previous data is required.

---

## Examples

```text
Filter

Map

Transform

Uppercase Conversion
```

---

## Stateless Diagram

```text
Record1
      ↓
 Process
      ↓
Output

Record2
      ↓
 Process
      ↓
Output
```

Each record is independent.

---

# What is Stateful Processing?

Requires remembering previously processed data.

---

## Examples

```text
Count Orders

Running Total

Aggregations

Joins

Windowing
```

---

## Stateful Diagram

```text
Record1
      ↓

State Store

      ↓

Record2

      ↓

Updated State
```

---

# Stateful vs Stateless Comparison

## Stateless

```text
No Memory Needed
```

Examples:

```text
Filter

Map
```

---

## Stateful

```text
Previous Data Required
```

Examples:

```text
Count

Aggregation

Join
```

---

# What is a State Store? ⭐⭐⭐

Kafka Streams stores intermediate state locally.

This storage is called:

```text
State Store
```

---

## What Does It Store?

Examples:

```text
Running Count

Revenue Totals

Latest Customer Data

Aggregation Results
```

---

# State Store Diagram

```text
Input Events
       ↓

Kafka Streams

       ↓

State Store

       ↓

Result
```

---

# What is Windowing? ⭐⭐⭐

Windowing allows processing records within a specific time period.

---

## Example

Business requirement:

```text
Sales Every 5 Minutes
```

Window:

```text
10:00 - 10:05
```

Sales:

```text
5000
```

---

# Why Windowing?

Data streams are infinite.

To calculate metrics:

```text
Per Minute

Per Hour

Per Day
```

we use windows.

---

# Tumbling Window

Fixed-size non-overlapping windows.

---

## Example

```text
10:00 - 10:05

10:05 - 10:10

10:10 - 10:15
```

No overlap.

---

## Diagram

```text
|-----|
10:00 10:05

|-----|
10:05 10:10

|-----|
10:10 10:15
```

---

# Hopping Window

Fixed-size windows with overlap.

---

## Example

Window Size:

```text
5 Minutes
```

Advance Every:

```text
2 Minutes
```

---

## Windows

```text
10:00 - 10:05

10:02 - 10:07

10:04 - 10:09
```

Overlap exists.

---

## Diagram

```text
|-----|
10:00 10:05

  |-----|
10:02   10:07

    |-----|
10:04     10:09
```
