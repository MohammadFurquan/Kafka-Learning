
# <span style="color:blue">ISR (In-Sync Replicas)</span>

## What is ISR?

ISR stands for:

```text
In-Sync Replicas
```

ISR is the set of replicas that are fully synchronized with the Leader replica.

Simple Definition:

```text
Replicas that are fully caught up with the Leader
=
ISR
```

---

## Why Do We Need ISR?

Kafka needs to know:

```text
Which replicas have the latest data?
```

If an outdated replica becomes the Leader:

```text
Data Loss Possible
```

ISR helps Kafka identify replicas that are safe for leader election.

---

# Replicas vs ISR

## Replica

A replica is simply a copy of a partition.

Example:

```text
Broker1 → Leader

Broker2 → Replica

Broker3 → Replica
```

All three are replicas.

---

## ISR

Only replicas that are fully synchronized with the Leader belong to ISR.

Example:

```text
Broker1 → Leader ✅

Broker2 → In Sync ✅

Broker3 → Lagging ❌
```

ISR:

```text
ISR = {Broker1, Broker2}
```

---

## Important Interview Point

```text
ISR ≠ All Replicas
```

A replica can exist but still not be part of ISR.

# Out-of-Sync Replica

Suppose Broker3 becomes slow because of a network issue.

Leader:

```text
Order1
Order2
Order3
Order4
Order5
```

Broker3:

```text
Order1
Order2
```

Broker3 is now behind.

Kafka removes it from ISR.

---

## New ISR

```text
ISR

{
 Broker1,
 Broker2
}
```

Broker3:

```text
Out-of-Sync Replica
```

---

# ISR Shrink

If a replica falls behind:

Before:

```text
ISR = {1,2,3}
```

After:

```text
ISR = {1,2}
```

This process is called:

```text
ISR Shrink
```

---

# ISR Expand

Suppose Broker3 catches up with the Leader.

Current:

```text
Order1
Order2
Order3
Order4
Order5
```

Broker3 synchronizes all records.

Now:

```text
ISR = {1,2,3}
```

This process is called:

```text
ISR Expand
```

---

# Leader Is Part of ISR ⭐

Important:

```text
Leader ∈ ISR
```

Example:

```text
ISR =
{
 Leader,
 Follower1,
 Follower2
}
```

The Leader is always considered part of ISR.

---

# ISR and Leader Election

Current State:

```text
Broker1 → Leader

Broker2 → ISR

Broker3 → Out Of Sync
```

Leader crashes:

```text
Broker1 ❌
```

Kafka prefers electing the new Leader from ISR.

Result:

```text
Broker2 → New Leader
```

---

## Interview Rule

```text
Kafka prefers leader election from ISR replicas.
```

# ISR and acks=all

Producer:

```text
acks = all
```

Flow:

```text
Producer
      ↓

Leader
      ↓

ISR Replicas
      ↓

ACK
```

Producer receives acknowledgement only after the required ISR replicas acknowledge the write.

# min.insync.replicas ⭐

Kafka provides:

```text
min.insync.replicas
```

This defines the minimum number of ISR replicas required for a successful write.

Example:

```text
Replication Factor = 3

acks = all

min.insync.replicas = 2
```

Kafka requires:

```text
Leader
+
At Least One More ISR Replica
```

Otherwise the write operation fails.


# <span style="color:blue">min.insync.replicas</span>

## What is min.insync.replicas?

`min.insync.replicas` defines the minimum number of **In-Sync Replicas (ISR)** that must be available for Kafka to successfully accept a write when using:

```text
acks=all
```

---

# Why Do We Need min.insync.replicas?

Kafka uses this setting to prevent unsafe writes and reduce the risk of data loss.

Without a minimum ISR requirement:

```text
Only 1 copy of data may exist
```

If that broker fails:

```text
Data Loss Possible
```

---

# Understanding the Components

## Replication Factor

```text
Replication Factor = 3
```

Means:

```text
Broker1 → Leader

Broker2 → Follower

Broker3 → Follower
```

Total Copies:

```text
3
```

---

## ISR

Suppose all replicas are synchronized.

```text
ISR = {Broker1, Broker2, Broker3}
```

ISR Count:

```text
3
```

---

## Configuration Example

```text
Replication Factor = 3

min.insync.replicas = 2

acks = all
```

Meaning:

```text
At least 2 ISR replicas must be available
for Kafka to accept writes.
```

# Why Does Kafka Reject The Write?

Kafka wants to ensure enough synchronized copies exist before accepting new data.

If Kafka allowed the write:

```text
Only Leader Has Data
```

and the Leader crashes:

```text
Data Lost
```

To avoid this risk, Kafka blocks writes when ISR falls below the minimum requirement.
