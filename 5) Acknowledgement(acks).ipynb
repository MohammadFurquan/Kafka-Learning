
# <span style="color:blue">Kafka Acknowledgements (acks)</span>

## What is Acknowledgement (ACK)?

When a Producer sends a message to Kafka, it needs confirmation that the message has been successfully stored.

This confirmation is called:

```text
ACK = Acknowledgement
```

Flow:

```text
Producer
    ↓
Kafka
    ↑
   ACK
```

Meaning:

```text
Message Successfully Received
```

---

# Why Do We Need ACK?

Suppose:

```text
Producer
    ↓
Kafka
```

Question:

```text
How does the producer know
the message was stored successfully?
```

Kafka sends an ACK.

Without ACK:

```text
Producer cannot know
whether the message was stored or lost.
```

---

# What is acks?

acks is a Kafka Producer configuration that controls how many Kafka replicas must confirm receiving a message before the Producer considers the write successful.

```text
acks
=
Acknowledgement Level
```

Kafka provides 3 modes:

```text
acks=0

acks=1

acks=all
```

---

# 1. acks = 0

## Meaning

Producer does not wait for any acknowledgement.

Flow:

```text
Producer
    ↓
Send Message

No ACK Wait
```

Producer sends the message and immediately continues.

---

## Diagram

```text
Producer
    ↓
 Leader

(No ACK Required)
```

---

## Risk

Suppose:

```text
Producer
    ↓
Message Sent
```

Network failure occurs.

```text
Message Lost
```

Producer will never know.

---

## Benefits

✅ Fastest Performance

✅ Lowest Latency

✅ Maximum Throughput

---

## Drawbacks

❌ Data Loss Possible

❌ No Reliability Guarantee

---

## Use Cases

```text
Application Logs

Monitoring Data

Metrics
```

Where small data loss is acceptable.

---

# 2. acks = 1

## Meaning

Producer waits only for the Leader replica to acknowledge.

Flow:

```text
Producer
    ↓
Leader
    ↓
 ACK
```

---

## Example

```text
Producer
    ↓
Broker1 (Leader)
```

Leader stores message and sends acknowledgement.

```text
ACK Received
```

Producer considers the write successful.

---

## Diagram

```text
Producer
    ↓

Leader
    ↓

 ACK
```

Followers may not have received the data yet.

---

## Risk

Scenario:

```text
Leader Receives Message
```

```text
ACK Sent To Producer
```

Immediately after:

```text
Leader Crashes
```

Before followers copy the data.

Result:

```text
Potential Data Loss
```

---

## Benefits

✅ Faster Than acks=all

✅ Better Reliability Than acks=0

✅ Lower Latency

---

## Drawbacks

❌ Data Loss Still Possible

---

# 3. acks = all ⭐

## Meaning

Producer waits for Leader and all ISR replicas to confirm the write.

Flow:

```text
Producer
      ↓

Leader
      ↓

ISR Replicas
      ↓

ACK
```

---

## Diagram

```text
Producer
    ↓

Leader

 ↓      ↓

F1      F2

    ↓

   ACK
```

Producer receives acknowledgement only after replication requirements are satisfied.

---

## Advantages

✅ Highest Reliability

✅ Highest Durability

✅ Best Data Safety

✅ Production Recommended

---

## Drawbacks

❌ Slightly Higher Latency

❌ More Network Communication

---

# Comparison of ACK Modes

## acks = 0

```text
Confirmation Required:
None
```

Reliability:

```text
Low
```

Speed:

```text
Fastest
```

---

## acks = 1

```text
Confirmation Required:
Leader Only
```

Reliability:

```text
Medium
```

Speed:

```text
Fast
```

---

## acks = all

```text
Confirmation Required:
Leader + ISR Replicas
```

Reliability:

```text
Highest
```

Speed:

```text
Slowest
```

---

# Real World Example

Architecture:

```text
MySQL
   ↓
Kafka
   ↓
Snowflake
```

For business-critical pipelines:

```properties
acks=all
```

Reason:

```text
Data Safety
>
Speed
```

---

# Important Failure Scenario

Before Failure:

```text
Producer
    ↓

Leader

 ↓      ↓

F1      F2
```

Producer sends message.

Leader stores message.

ACK is sent.

Immediately:

```text
Leader Crashes
```

Question:

Which ACK mode was used?

Answer:

```text
acks=1
```

Because only Leader acknowledgement was required.

# <span style="color:red">Acknowledgements (acks) Interview Questions & Answers</span>

## Q1. What is ACK in Kafka?

ACK (Acknowledgement) is a confirmation sent by Kafka to the Producer indicating that a message has been successfully received.

---

## Q2. Why Do We Need ACKs?

ACKs help the Producer know whether the message was successfully stored or lost.

---

## Q3. What Does acks Mean in Kafka?

acks is a Producer configuration that determines how many replicas must acknowledge a write before it is considered successful.

---

## Q4. What Are the Available ACK Modes?

```text
acks=0

acks=1

acks=all
```

---

## Q5. What Does acks=0 Mean?

Producer does not wait for any acknowledgement.

---

## Q6. What Is the Benefit of acks=0?

✅ Fastest Performance

✅ Lowest Latency

---

## Q7. What Is the Risk of acks=0?

❌ Data Loss Possible

---

## Q8. What Does acks=1 Mean?

Producer waits for acknowledgement from the Leader replica only.

---

## Q9. What Is the Benefit of acks=1?

✅ Faster than acks=all

✅ More reliable than acks=0

---

## Q10. What Is the Risk of acks=1?

Leader may crash before followers replicate the data, causing data loss.

---

## Q11. What Does acks=all Mean?

Producer waits for acknowledgement from Leader and ISR replicas.

---

## Q12. Which ACK Mode Provides Maximum Reliability?

```text
acks=all
```

---

```text
acks=0
```

---

## Q14. Which ACK Mode Is Safest?

```text
acks=all
```

```text
acks=all
```

## Q16. Which ACK Mode Has The Lowest Latency?

```text
acks=0
```

## Q17. Which ACK Mode Offers A Balance Between Speed And Reliability?
