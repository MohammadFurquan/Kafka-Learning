# <span style="color:green">What is Kafka?</span>

Apache Kafka is an open-source distributed event streaming platform that receives data from producers, stores it reliably, and delivers it to consumers in real time.

Kafka is:

✅ Distributed

✅ Fault-Tolerant

✅ Scalable

✅ High Throughput

✅ Real-Time

✅ Event Driven

### Simple Definition

Kafka stores data sent by Producers and delivers it to Consumers.

```text
Producer
    ↓
 Kafka
    ↓
Consumer
```

---

# <span style="color:green">Need for Apache Kafka</span>

Modern applications generate huge amounts of real-time data from:

- Websites
- Mobile Applications
- IoT Devices
- Transactions
- User Activities

Traditional systems struggle to process such large-scale data efficiently.

Kafka solves these problems by providing:

### Real-Time Processing

Processes live data streams instantly.

### Fault Tolerance

Replicates data across brokers and prevents data loss.

### Scalability

Supports horizontal scaling and handles growing workloads.

### Event-Driven Architecture

Systems react automatically to events.

### High Throughput

Processes millions of messages per second.

### Offset Management

Consumers can continue reading from the last saved position.

# <span style="color:green">Core Components</span>

## 1. Kafka Broker

A Broker is a Kafka Server responsible for storing and managing messages.

### Responsibilities

- Receive messages
- Store messages
- Send messages to consumers
- Handle producer requests
- Handle consumer requests
- Support scalability
- Support fault tolerance

### Interview Definition

```text
Broker = Single Kafka Server
```

---

## 2. Producer

A Producer is an application that sends messages to Kafka Topics.

### Producer can send:

- Logs
- Transactions
- Events
- Metrics
- User Activities

### Example

```text
Order Created

Payment Success

User Login
```

Producer also decides how data is distributed across partitions.

---

## 3. Kafka Topic

A Topic is a category where messages are stored.

### Example

```text
orders

payments

customers
```

### Topic Characteristics

- Producers write to Topics
- Consumers read from Topics
- Every message belongs to a Topic
- Topics are divided into Partitions

---

## 4. Consumer

A Consumer reads messages from Kafka Topics.

### Example

```text
Analytics Service

Email Service

Fraud Detection Service
```

---

## 5. Consumer Group

A Consumer Group is a collection of consumers reading from the same topic.

### Benefits

✅ Load Balancing

✅ Parallel Processing

✅ Scalability

✅ Fault Tolerance

---

### Golden Rule

```text
Within the same Consumer Group

1 Partition = 1 Consumer
```

A partition cannot be consumed by multiple consumers of the same consumer group simultaneously.

---

### Example

```text
Orders Topic

P0
P1
P2
```

Consumer Group:

```text
C1
C2
C3
```

Assignment:

```text
P0 → C1

P1 → C2

P2 → C3
```

---

## 6. ZooKeeper

ZooKeeper was used to manage Kafka clusters.

### Responsibilities

- Broker Coordination
- Metadata Management
- Leader Election
- Cluster Synchronization
- Failure Recovery

### Note

Modern Kafka is moving towards KRaft mode and reducing dependency on ZooKeeper.

# <span style="color:green">Kafka Cluster</span>

In real-world systems, we do not use a single Broker.

Multiple Brokers together form a Kafka Cluster.

```text
Broker1

Broker2

Broker3
```

```text
Kafka Cluster
```

---

## Why Do We Need Clusters?

A single broker can become overloaded.

Clusters provide:

✅ Scalability

✅ Fault Tolerance

✅ High Availability

---

## Broker vs Cluster

```text
Broker = Single Kafka Server

Cluster = Collection of Kafka Brokers
```

# <span style="color:green">Important Kafka Concepts</span>

## Node

A Node refers to an individual server within a Kafka Cluster.

```text
Node = Broker
```

## Replica

### Important

# <span style="color:green">Workflow of Apache Kafka</span>

## Step 1: Producer Sends Data

```text
Producer
    ↓
 Topic
```

Data can be:

- Logs
- Transactions
- Events
- Metrics

---

## Step 2: Kafka Stores Data

Kafka stores data inside Topics.

Messages remain available for a configured retention period.

Data is replicated across brokers.

---

## Step 3: Consumer Reads Data

Consumers subscribe to Topics.

Consumer Groups distribute workload.

---

## Step 4: Kafka Balances Load

Kafka distributes Partitions across Brokers.

If a broker fails:

```text
Replica takes over automatically
```

---

## Step 5: Data Processing

Consumers process data for:

- Analytics
- Storage
- Monitoring
- Alerts
- Notifications

Kafka integrates with:

- Spark
- Flink
- Hadoop

# <span style="color:green">Offset</span>

Offset is a unique sequence number assigned to each message within a partition.

### Example

```text
Partition-0

Offset 0 → Order1

Offset 1 → Order2

Offset 2 → Order3

Offset 3 → Order4
```

---

## Why Offset?

Kafka uses offset to track message position.

```text
Book → Page Number

Kafka → Offset
```

---

## Important Point

Offsets are unique only within a partition.

Example:

```text
Partition-0

0
1
2
3
```

```text
Partition-1

0
1
2
3
```

Offsets can repeat across partitions.

---

# <span style="color:green">Consumer Offset</span>

Consumer Offset shows how far a consumer has read data.

### Example

```text
Partition

0
1
2
3
4
5
```

Consumer processed:

```text
0
1
2
3
```

Kafka stores:

```text
Consumer Offset = 3
```

---

## Consumer Restart Example

Consumer crashes.

Kafka remembers:

```text
Consumer Offset = 3
```

Consumer resumes from:

```text
Offset 4
```

---

## Benefits

✅ Recovery After Crash

✅ Resume Processing

✅ Replay Support

✅ No Data Loss

---

## Difference

```text
Offset
=
Message Position
```

```text
Consumer Offset
=
Consumer Last Read Position
```

# <span style="color:green">Replication</span>

Replication means creating multiple copies of a partition.

---

## Why Replication?

Suppose:

```text
Broker1 Crash
```

Without replication:

```text
Data Lost
```

With replication:

```text
Broker2

Broker3
```

Data remains available.

---

## Replication Factor

Example:

```text
RF = 3
```

Means:

```text
1 Leader

2 Followers
```

Total Copies:

```text
3
```

---

## Benefits

✅ Fault Tolerance

✅ High Availability

✅ Data Protection

✅ Disaster Recovery

---

# <span style="color:green">Leader-Follower Architecture</span>

Each partition has one Leader and multiple Followers.

### Example

```text
Partition-0

Broker1 → Leader

Broker2 → Follower

Broker3 → Follower
```

---

## Producer Writes To

```text
Leader Only
```

---

## Consumer Reads From

```text
Leader Only
```

(Generally)

---

## Follower Responsibility

Followers continuously synchronize data with the Leader.

---

## Leader Failure

Example:

```text
Leader = Broker1
```

Broker1 crashes.

Kafka automatically promotes:

```text
Broker2 → New Leader
```

---

## Benefits

✅ High Availability

✅ Continuous Processing

✅ Automatic Failover

✅ No Downtime

# <span style="color:green">Retention</span>

Kafka stores messages even after they are consumed.

---

## Traditional Queue

```text
Read Message
      ↓
 Deleted
```

---

## Kafka

```text
Read Message
      ↓
 Still Stored
```

---

## Example

```text
Retention = 7 Days
```

Consumer reads message.

Message still remains available for 7 days.

---

## Why Retention?

### Replay

Read old messages again.

### Backfill

New consumer can read historical data.

### Audit

View historical events.

---

## Retention Types

### Time Based

```text
7 Days

14 Days

30 Days
```

---

### Size Based

```text
100 GB

500 GB

1 TB
```

When size limit is reached:

```text
Old data gets deleted
```

---

## Benefits

✅ Replay

✅ Recovery

✅ Backfill

✅ Audit Trail

✅ Historical Analysis

## Q12. What happens if a Consumer crashes?

**Answer:**

Kafka stores the consumer offset.

After restart, the consumer continues from the next offset.

Example:

```text
Last Offset = 100

Restart

Continue From = 101
```

## Q18. What is a Leader Replica?

**Answer:**

Leader Replica handles:

✅ All Writes

✅ Most Reads

---

## Q19. What is a Follower Replica?

**Answer:**

Follower Replicas continuously copy data from the Leader Replica.

Purpose:

✅ Fault Tolerance

✅ High Availability

✅ Data Protection

---

## Q20. What happens if the Leader fails?

**Answer:**

Kafka automatically promotes one of the followers to become the new Leader.

Example:

Before:

```text
Broker1 → Leader

Broker2 → Follower

Broker3 → Follower
```

After Broker1 failure:

```text
Broker2 → Leader

Broker3 → Follower
```
