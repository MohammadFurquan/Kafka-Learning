
# <span style="color:blue">Serialization & Deserialization in Apache Kafka</span>

# What is Serialization?

Serialization is the process of converting data, objects, or messages into bytes before sending them to Kafka.

### Simple Definition

```text
Object / Data
      ↓
Serialization
      ↓
Bytes
```

Kafka can only transmit and store:

```text
byte[]
```

Therefore, every message must be serialized before it is written to Kafka.

---

# Why Do We Need Serialization?

Suppose a Producer creates:

```json
{
  "customer_id": 101,
  "name": "Mohammad Arif"
}
```

Kafka cannot directly store this object.

It must first be converted into:

```text
Bytes
```

This conversion is called:

```text
Serialization
```

---

# What is Deserialization?

Deserialization is the reverse process.

It converts bytes back into the original data structure.

### Simple Definition

```text
Bytes
   ↓
Deserialization
   ↓
Object / Data
```

---

# Why Do We Need Deserialization?

Consumer receives:

```text
Bytes
```

To work with the data, the Consumer converts it back into:

```text
JSON

String

Java Object

Avro Object
```

This process is called:

```text
Deserialization
```

---

# End-to-End Flow

```text
Producer
    ↓
Serializer
    ↓
Kafka (Bytes)
    ↓
Deserializer
    ↓
Consumer
```

---

# Kafka Stores Bytes, Not Objects ⭐

Kafka does not understand:

```text
Java Objects

Python Objects

JSON Objects

Avro Objects
```

Kafka stores only:

```text
byte[]
```

Everything sent to Kafka is ultimately converted into bytes.

---

# How Kafka Stores Data On Disk

Important Interview Point ⭐

Apache Kafka does not use high-level file formats like CSV or Parquet for on-disk storage. Instead, it treats messages as raw byte arrays and stores them in binary log segments (.log files) on disk, organized into batches called Record.

---

# On-Disk Storage Structure ⭐

On-Disk Storage Structure

Log Segments (.log):
The actual binary data payload of messages.

Index Files (.index):
Maps logical offsets to physical file positions for fast lookups.

Time Index (.timeindex):
Maps timestamps to offsets for time-based queries.

---

# Kafka Storage Architecture

```text
Topic
  ↓

Partition
  ↓

000000000.log
000000000.index
000000000.timeindex

000000001.log
000000001.index
000000001.timeindex
```

---

# Purpose of Each File

## .log File

Stores:

```text
Actual Messages
```

Example:

```text
Order1

Order2

Order3
```

Stored as binary records.

---

## .index File

Stores:

```text
Offset Mapping
```

Example:

```text
Offset 100 → File Position X

Offset 200 → File Position Y
```

This enables fast lookups.

---

## .timeindex File

Stores:

```text
Timestamp → Offset
```

Example:

```text
10:00 AM → Offset 100

11:00 AM → Offset 500
```

Useful for time-based retrieval.


# <span style="color:blue">When Does Serialization and Deserialization Happen in Kafka?</span>

# Serialization on Producer Side

Serialization happens when the Producer sends a message to Kafka.

Example:

```java
producer.send(record);
```

At this point Kafka cannot directly send:

```text
Java Objects

Python Objects

JSON Objects

Avro Objects
```

Kafka can only send:

```text
byte[]
```

Therefore Kafka internally invokes the configured Serializer.

---

## Producer Flow

```text
Application Object
        ↓
 producer.send()
        ↓
 Serializer Runs
        ↓
 Object → Bytes
        ↓
 Kafka Broker
```

---

## Example

Application Object:

```json
{
  "id": 101,
  "name": "Arif"
}
```

Producer:

```java
producer.send(record);
```

Kafka Internally:

```text
JSON Object
      ↓
JSON Serializer
      ↓
Bytes
      ↓
Kafka Broker
```

---

## Interview Point

```text
Serialization happens before data enters Kafka.
```

---

# Deserialization on Consumer Side

Deserialization happens when the Consumer reads data from Kafka.

Example:

```java
consumer.poll(...)
```

Kafka stores data as:

```text
Bytes
```

But applications need:

```text
Objects
```

Therefore Kafka internally invokes the configured Deserializer.

---

## Consumer Flow

```text
Kafka Topic
     ↓
Bytes
     ↓
consumer.poll()
     ↓
Deserializer Runs
     ↓
Bytes → Object
     ↓
Application Code
```

---

## Example

Kafka Record:

```text
Bytes
```

Consumer:

```java
consumer.poll(...)
```

Kafka Internally:

```text
Bytes
    ↓
JSON Deserializer
    ↓
JSON Object
```

Application Receives:

```json
{
  "id": 101,
  "name": "Arif"
}
```

---

# Important Interview Point ⭐

By the time your application gets:

```java
ConsumerRecord<String, Customer>
```

deserialization has already happened.

Flow:

```text
Kafka Fetches Bytes
        ↓
Deserializer Runs
        ↓
ConsumerRecord Created
        ↓
Your Code Receives Object
```
